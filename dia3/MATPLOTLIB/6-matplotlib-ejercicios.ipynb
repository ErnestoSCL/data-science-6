```python
import pandas as pd
df_books = pd.read_csv('https://github.com/cesarmayta/datasets/raw/refs/heads/main/libros.csv')
df_books.head(2)
```

```python
df_books.describe(include='object')
```

📌 **Ejercicio 1: Evolución del precio promedio por año**

Muestra cómo ha cambiado el precio promedio de los libros en el tiempo.

```python
import matplotlib.pyplot as plt
```

```python
# x = Year
# y = Precio Promedio
```

```python
df_eje1 = df_books.groupby('Year')['Price'].mean().reset_index()
df_eje1
```

```python
x = df_eje1['Year']
y = df_eje1['Price']
```

```python
# grafico
plt.plot(x, y, color='green', label='Precio Promedio')
plt.title('Evolución del precio promedio por año')
plt.xlabel('Año')
plt.ylabel('Precio Promedio')
plt.legend()
plt.show()
```

📌 **Ejercicio 2: Comparar calificación(Rating) promedio por género**
Compara la calificación promedio entre libros de ficción y no ficción, mostrar un grafico de barras

```python
df_eje2 = df_books.groupby('Genre')['User Rating'].mean().reset_index()
df_eje2
```

```python
# Comparar calificación(Rating) promedio por género Compara la calificación promedio entre libros de ficción y no ficción, mostrar un grafico de barras
x = df_eje2['Genre']
y = df_eje2['User Rating']
plt.bar(x,y,color=['blue','red'],label='comparación por genero')
plt.title('Calificación promedio por Genero')
plt.xlabel('Genero')
plt.ylabel('Calificación Promedio')
plt.legend()
plt.show()
```
